## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
campos = ['Secuencial', 'Fecha de Análisis', 'Departamento', 'Municipio', 'pH agua:suelo']
print(campos)
print("=====================================================")

#Se suman los nulos
nulos = suelo[campos].isna().sum()
print(nulos)
print("=====================================================")

#Se aplica ciclo para contar vacios
vacios = suelo[campos].apply(lambda col: col.str.strip().eq("")).sum()
print(vacios)
print("=====================================================")

#Se cuentan nd haciendo transformaciones de texto para matchear
nd = suelo[campos].apply(lambda col: col.str.strip().str.upper().eq("ND")).sum()
print(nd)
print("=====================================================")

# Crear y mostrar perfil con las cinco columnas solicitadas.
perfil = pd.DataFrame({
    "campo": campos,
    "nulos_pandas": nulos.values,
    "vacios_tras_strip": vacios.values,
    "marcador_ND": nd.values,
})
display(perfil)
print("=====================================================")

#comprobación
assert len(perfil) == 5, f"Se esperaban 5 filas y hay {len(perfil)}"
print("Perfil tiene una fila por campo")
````

**Interpretación del resultado:**

La tabla `perfil` resume tres tipos de dato faltante en cinco campos. Ningún campo tiene nulos de pandas, lo cual es consistente con la lectura usando `keep_default_na=False`: los vacíos y los marcadores no se convierten automáticamente en `NaN`, por lo que `isna()` por sí solo no detecta nada.

Al contar los textos vacíos después de retirar espacios, aparece un único hallazgo: `Municipio` tiene **2 registros vacíos**. `Secuencial`, `Fecha de Análisis`, `Departamento` y `pH agua:suelo` no tienen vacíos, y ninguno de los cinco campos contiene el marcador `ND`.

3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
ph_texto = suelo["pH agua:suelo"].str.strip().str.replace(",", ".", regex=False)
suelo["ph"] = pd.to_numeric(ph_texto, errors="coerce")
print(suelo["ph"].describe())
print("=====================================================")

#Convertir fecha
fecha_texto = suelo["Fecha de Análisis"].str.strip()
suelo["fecha_dt"] = pd.to_datetime(fecha_texto, format="%d/%m/%Y", errors="coerce")
print(suelo["fecha_dt"].min(), suelo["fecha_dt"].max())
print("=====================================================")

#Vacias y fallidos
ph_vacio = suelo["pH agua:suelo"].str.strip().eq("")
ph_fallido = suelo["ph"].isna() & ~ph_vacio

fecha_vacia = suelo["Fecha de Análisis"].str.strip().eq("")
fecha_fallida = suelo["fecha_dt"].isna() & ~fecha_vacia
print("=====================================================")

#Conteos
print("pH vacíos:", ph_vacio.sum())
print("pH fallidos:", ph_fallido.sum())
print("Fechas vacías:", fecha_vacia.sum())
print("Fechas fallidas:", fecha_fallida.sum())
print("=====================================================")

#Fallidos
print("Ejemplos pH fallidos:")
display(suelo.loc[ph_fallido, ["Secuencial", "pH agua:suelo", "ph"]].head(5))

print("Ejemplos fechas fallidas:")
display(suelo.loc[fecha_fallida, ["Secuencial", "Fecha de Análisis", "fecha_dt"]].head(5))
print("=====================================================")

# Mostrar los fallos de conversión y sus valores originales.
assert len(suelo) == filas_originales
assert suelo["ph"].isna().sum() == ph_vacio.sum() + ph_fallido.sum()
assert suelo["fecha_dt"].isna().sum() == fecha_vacia.sum() + fecha_fallida.sum()
print("✅ Sin filas perdidas; cada NaN/NaT es vacío o fallido; no se rellenó nada con cero")
````

**Interpretación del resultado:**

Se crearon `ph` y `fecha_dt` conservando las columnas originales. La tabla mantiene sus 92.738 filas.

En `pH agua:suelo` no hay entradas vacías, pero **11 textos no vacíos no pudieron convertirse**. Los ejemplos muestran dos causas: un error de digitación (`4..01`) y valores censurados del laboratorio (`>10.00`, `<3.80`), que indican un límite de detección y no una medida exacta. Estos casos quedan como `NaN` en `ph`; no se reemplazaron por cero ni por otro valor, porque eso introduciría medidas falsas.

En `Fecha de Análisis` no hubo vacíos ni conversiones fallidas: las 92.738 fechas se interpretaron con el formato `%d/%m/%Y`

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
motivo_ph_ausente = suelo["ph"].isna()
print("pH ausente:", motivo_ph_ausente.sum())

motivo_ph_fuera = suelo["ph"].notna() & ~suelo["ph"].between(0, 14)
print("pH fuera de [0, 14]:", motivo_ph_fuera.sum())

motivo_fecha = suelo["fecha_dt"].isna()
print("Fecha no interpretada:", motivo_fecha.sum())

suelo["en_revision"] = motivo_ph_ausente | motivo_ph_fuera | motivo_fecha
print(suelo["en_revision"].value_counts())

resumen = pd.DataFrame({
    "motivo": ["pH ausente", "pH fuera de [0, 14]", "Fecha no interpretada", "Total filas marcadas"],
    "filas": [motivo_ph_ausente.sum(), motivo_ph_fuera.sum(), motivo_fecha.sum(), suelo["en_revision"].sum()],
})
display(resumen)

# Comprobar la conservación del total original con assert.
marcadas = suelo["en_revision"].sum()
no_marcadas = (~suelo["en_revision"]).sum()
assert len(suelo) == filas_originales
assert marcadas + no_marcadas == filas_originales
print("Marcadas + no marcadas =", marcadas + no_marcadas)
````

**Interpretación del resultado:**

Se creó la columna booleana `en_revision`, que marca una fila si su pH es ausente, está fuera de [0, 14] o su fecha no se pudo interpretar. No se eliminó ninguna observación: las 11 filas marcadas y las 92.727 no marcadas suman las 92.738 originales.

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
controles = pd.DataFrame({
    "control": [
        "Total de entrada",
        "pH sin número",
        "pH fuera del intervalo",
        "Fecha no interpretable",
        "Total de filas en revisión",
    ],
    "cantidad": [
        filas_originales,
        int(motivo_ph_ausente.sum()),
        int(motivo_ph_fuera.sum()),
        int(motivo_fecha.sum()),
        int(suelo["en_revision"].sum()),
    ],
})
display(controles)
#=================================================

ruta_controles = OUT / "controles_agrosavia.csv"
controles.to_csv(ruta_controles, index=False)
print("Archivo generado:", ruta_controles)
#=================================================

assert ruta_controles.is_file()
assert ruta_controles.parent == OUT
print("Archivo dentro de OUT")
display(pd.read_csv(ruta_controles))
# Redactar los dos hallazgos en la celda Markdown siguiente.
````

**Interpretación del resultado:**

La tabla `controles` resume en cinco conteos la calidad de la entrada y se exportó sin índice a `OUT / "controles_agrosavia.csv"`. De las **92.738** filas de entrada, **11** tienen un pH sin número, **0** tienen pH fuera de [0, 14], **0** tienen fecha no interpretable y **11** quedan en revisión. Ninguna fila se eliminó.

**Hallazgos:**

1. **Casi todas las filas son utilizables para pH y fecha.** Solo **11** de 92.738 filas tienen un pH que no se pudo convertir, y las fechas se interpretaron todas con `%d/%m/%Y`. Esto significa que la tabla puede usarse para análisis de pH y de tiempo casi completa, pero esas 11 filas no deben entrar a cálculos numéricos sin una decisión previa.

2. **Los 11 casos de pH sin número son textos con significado, no datos ausentes en origen.** Los ejemplos corresponden a valores censurados del laboratorio y a un error de digitación. Como hay **0** pH fuera de [0, 14], el total en revisión coincide con esos 11. La limitación es que los controles solo detectan lo que no se convierte o cae fuera del rango: un pH numérico plausible pero erróneo, dentro de [0, 14], no queda marcado.
